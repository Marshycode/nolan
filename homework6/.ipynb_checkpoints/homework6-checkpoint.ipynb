```python
arra = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
def primey(numbert):
    if numbert<2:
        return False
    for i in range(2,int(numbert**0.5)+1):
                if numbert % i == 0:
                    return False
    return True
    
def alien_prediction(arra):
    mask = np.array([any(primey(x) for x in row)for row in arra])
    return arra[mask]
alien_prediction(arra)
```

```python

x=[]
x=np.zeros((8,8), dtype=int) 
print (x)
    
```

```python
def create_board():
    boarding=np.array([1 if i % 2==0 else 0 for i in range(8)])
    np.array(x)
    x[1::2] = boarding
    return x
board = []
board = create_board()
print(board)
```

```python

def rev_create_board():
    boardi=np.array([0 if i % 2==0 else 1 for i in range(8)])
    np.array(x)
    x[1::2] = boardi
    return x
rev_board = []
rev_board = rev_create_board()
print(rev_board)
```

```python
def spaces(text, num_spaces):
    spaces = ' '*num_spaces
    return spaces.join(text) 
print (spaces("galaxy",2))
```
