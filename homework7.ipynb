```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x,A,w):
    y = A* np.sin(w*x) 
    return y

x = np.linspace(0, 2 * np.pi, 1000) 
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
fig, axes = plt.subplots(1,3,figsize=(25,5)) 
axes[0].set_title("Subplot 1: V-Amplitued") 
axes[1].set_title("Subplot 2: V-Frequency") 
axes[2].set_title("Subplot 3: Amplitude Frequency") 
for A,w in zip(amplitude, frequency):
    print (f"Amplitude: {A}, Frequency: {w}") 
for A in amplitude:
    y=make_sine_wave(x,A,2)
    axes[0].plot(x,y)
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].legend()
axes[0].grid(True) 
for w in frequency:
    y=make_sine_wave(x,1,w)
    axes[1].plot(x,y)
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")
axes[1].legend()
axes[1].grid(True) 
for A,w in zip(amplitude, frequency):
    y=make_sine_wave(x,A,w)
    axes[2].plot(x,y)
axes[2].set_xlabel("x")
axes[2].set_ylabel("y")
axes[2].legend()
axes[2].grid(True) 

plt.show()

```
