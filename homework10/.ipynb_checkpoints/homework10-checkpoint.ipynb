```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
df=pd.read_csv('solar_system.csv')
```

```python
df.shape
print (df) 
```

```python
df = df.set_index("Attribute").T
df.index.name = "Planet"
df.reset_index(inplace=True)
df.columns.name = None
```

```python
print (df)
df.shape
```

```python
#There are rows 10 and 21 columns. The code first finds the attributes column, then changed the index to planet, then reset the index by switching the column to planets.
#The shape removed a row and added it to the columns.
print (df.columns)
```

```python
print(df.dtypes)
for col in df.columns:
    print(f"Column: {col}")
    print(df[col].apply(type).unique())

#Currently, they are strings. The issue of storing data as strings mean they are much less workable as they connot be easily manipulated.
```

```python
print(df.dtypes)
exclude_cols = ["Planet", "Ring System?", "Global Magnetic Field?"]
for col in df.columns:
    if col not in exclude_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")
print(df.dtypes)
```

```python
#There are floats, ints and objects, floats contain decimals, ints are numbers without decimals, and objects wich are strings/its a term for many things in python.
#Planet, rings, and magnetic feild are all objects. 
```

```python
def attach_units(col_name, unit, new_col_name):
    new_values = []
    for val in df[col_name]:
        new_values.append(val * unit)
    df[col_name] = new_values
    df.rename(columns={col_name: new_col_name}, inplace=True)
unit = 1e24 * u.kg 
attach_units("Mass (10^24kg)", unit, "Mass (kg)")
print(df["Mass (kg)"])
```

```python
peri_col = "Perihelion (10^6 km)"
aphe_col = "Aphelion (10^6 km)"
raw_semimajor = (df[peri_col] + df[aphe_col]) / 2
unit_dist = 1e6 * u.km
semimajor_with_units = [val * unit_dist for val in raw_semimajor]
loc_index = df.columns.get_loc(aphe_col) + 1
new_col_name = "Semi-major Axis (km)"

df.insert(loc_index, new_col_name, semimajor_with_units)
```

```python
print(df.columns)
```

```python
old_col = "Orbital Period (days)"
new_col = "Orbital Period (years)"
new_values = []
for val in df[old_col]:
    qty_days = val * u.d
    qty_years = qty_days.to(u.yr)
    new_values.append(qty_years)
df[old_col] = new_values
df.rename(columns={old_col: new_col}, inplace=True)
```

```python
print(df[df["Planet"] == "Mars"].iloc[0])
```

```python
#mars has an orbital period of 1.9 years
#AU is a distance from the earth to the sun
print (const.au)
```

```python
#149597870700.0 meters in one AU
dist_cols_config = [
    {"name": "Diameter (km)", "factor": 1.0, "is_quantity": False, "new_name": "Diameter (AU)"},
    {"name": "Distance from Sun (10^6 km)", "factor": 1e6, "is_quantity": False, "new_name": "Distance from Sun (AU)"},
    {"name": "Perihelion (10^6 km)", "factor": 1e6, "is_quantity": False, "new_name": "Perihelion (AU)"},
    {"name": "Aphelion (10^6 km)", "factor": 1e6, "is_quantity": False, "new_name": "Aphelion (AU)"},
    {"name": "Semi-major Axis (km)", "factor": 1.0, "is_quantity": True, "new_name": "Semi-major Axis (AU)"}
]
converted_cols = []

for config in dist_cols_config:
    col = config["name"]
    new_col = config["new_name"]
    converted_cols.append(new_col)
    
    new_values = []
    for val in df[col]:
        if config["is_quantity"]:
            qty_km = val
        else:
            qty_km = val * config["factor"] * u.km
            
        qty_au = qty_km.to(u.AU)
        new_values.append(qty_au)
    
df[col] = new_values
df.rename(columns={col: new_col}, inplace=True)
print (df[df["Planet"] == "Mars"].iloc[0])
```
