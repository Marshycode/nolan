```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
print(data.shape) 
```

```python
magnitude= data[:,0]
temperature= data[:,1]
print ("Minimum Magnitude:", magnitude.min())
print ("Maximum Magnitude:", magnitude.max())
print ("Minimum Temperature:", temperature.min())
print ("Maximum Temperature:", temperature.max())
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
brightest_idx = np.argmin(magnitude)
dimmest_idx = np.argmax(magnitude)
hottest_idx = np.argmax(temperature)
coldest_idx = np.argmin(temperature)
plt.figure(figsize=(8,6)) 
plt.scatter(log_temperature, log_magnitude, color="blue", label="stars") 
plt.scatter(log_temperature[brightest_idx],log_magnitude[brightest_idx], color="black", marker = "*", s = 300, label="Brightest Star") 
plt.scatter(log_temperature[dimmest_idx],log_magnitude[dimmest_idx], color="purple", marker = "*", s = 300, label="Dimmest Star")  
plt.scatter(log_temperature[hottest_idx],log_magnitude[hottest_idx], color="green", marker = "*", s = 300, label="Hottest Star")
plt.scatter(log_temperature[coldest_idx],log_magnitude[coldest_idx], color="orange", marker = "*", s = 300, label="Coldest Star")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.xlabel("Temperature (log K)") 
plt.ylabel("Magnitude")
plt.title("Hertzsprung-Russell Diagram") 
plt.legend()
plt.show()
        
```
