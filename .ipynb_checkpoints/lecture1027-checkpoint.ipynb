```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.optimize import curve_fit

df = pd.read_csv("xy_data.csv")
df
df.head()
```

```python
x=df["x"]
y=df["y"]
plt.plot(x,y, 'b+', label="data")
plt.title('Data to Fit')
plt.xlabel('x values')
plt.ylabel('y values')
plt.legend()
plt.show()
```

```python
#data is liner so fit y=mx+b
def linear_fit(x,m,b):
    return (m*x)+b
popt, pcov = curve_fit(linear_fit, x, y) 
m_fit, b_fit =popt
print(f"Line of Best Fit: y = {m_fit:.2f}x + {b_fit:.2f}")
```

```python
x=df["x"]
y=df["y"]
plt.plot(x,y, 'b+', label="data")
plt.plot(x,m_fit*x + b_fit, color="red", label="Linear Fit")
plt.title('Data to Fit')
plt.xlabel('x values')
plt.ylabel('y values')
plt.legend()
plt.show()
```

```python
#error
perr=np.sqrt(np.diag(pcov))
m_err, b_err =perr

print(f"Slope: {m_fit:.2f} +/- {m_err:.2f}")
```

```python
#Reduced chi-squared
rediduals = y - linear_fit(x, m_fit, b_fit)
chi2 = np.sum(residuals**2)
dof = len(x)-len(popt)
chi2_squared = chi2/ dof
print(f"reduced chi-squared: {chi2_squared:.2f}")
```
