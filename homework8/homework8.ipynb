```python
med1=df['latitude'].median()
med2=df['longitude'].median()
print(med1,med2) 
```

```python
df["Dist_to Med_Longitude"]=(df["longitude"]-med2).abs()
print (df["Dist_to Med_Longitude"].head())
```

```python
np.shape(df)
#52 rows and five columns
```

```python
df["Dist_to Med_latitude"]=(df["latitude"]-med1).abs()
close_lat_state=df.loc[df["Dist_to Med_latitude"].idxmin()]
far_lat_state=df.loc[df["Dist_to Med_latitude"].idxmax()]
close_long_state=df.loc[df["Dist_to Med_Longitude"].idxmin()]
far_long_state=df.loc[df["Dist_to Med_Longitude"].idxmax()]

print (f"closest to med lat:{close_lat_state['state']} ({close_lat_state['Dist_to Med_latitude']:.4f}away)")
print (f"farthest to med lat:{far_lat_state['state']} ({far_lat_state['Dist_to Med_latitude']:.4f}away)")
print (f"closest to med lat:{close_long_state['state']} ({close_long_state['Dist_to Med_Longitude']:.4f}away)")
print (f"farthest to med lat:{far_long_state['state']} ({far_long_state['Dist_to Med_Longitude']:.4f}away)")
       
```

```python
geyser = sns.load_dataset("geyser")
print (df)
```

```python
geyser1 = geyser.dropna(axis=0) 
print (df)
```

```python
#272x3, 272 rows and 3 columns
#printing automatically calls first five and last five rows.
#column one is eruption number, two is geyser eruption time and waiting is time between eruption.
geyser2 = geyser1.dropna(axis=1) 
print (df)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

eruptions = geyser2.iloc[0]

sns.scatterplot(
    x="duration",
    y="waiting",
    data = geyser2,
    ax=axes[0,0],
    palette="cool"
)

axes[0, 0].set_title("Eruption Duration vs Waiting Time")
axes[0, 0].set_xlabel("Eruption Duration (m)")
axes[0, 0].set_ylabel("Waiting Time (m)")

sns.scatterplot(
    x="waiting",
    y="duration",
    data=geyser2,
    ax=axes[0, 1],
)
axes[0, 1].set_title("Waiting Time vs Eruption Duration")
axes[0, 1].set_xlabel("Waiting Time (minutes)")
axes[0, 1].set_ylabel("Eruption Duration (minutes)")

sns.histplot(
    data=geyser2,
    x="kind",
    bins=20,
    kde=True,
    color="green",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Distribution of Eruption Durations")
axes[1, 0].set_xlabel("Short or Long")
axes[1, 0].set_ylabel("Count")

sns.boxplot(
    data=geyser2,
    x="kind",
    y="waiting",
    palette="Set2",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Waiting Time by Eruption Kind")
axes[1, 1].set_xlabel("Eruption Kind")
axes[1, 1].set_ylabel("Waiting Time (minutes)")

plt.tight_layout()
plt.show() 
```
